```python
#路径path
from pathlib import Path

path = Path('pi_digits.txt')#分相对路径和绝对路径
content = path.read_text()
print(content)
```

```python
#多返回了一个空字符串
content1 = content.rstrip()
print(content1)
```

```python
content2 = content.splitlines()#分行
print(content2)
```

```python
pi_string = ''
for line in content2:
    pi_string += line
print(pi_string)
print(len(pi_string))
```

```python
from pathlib import Path
p1 = Path('programing.txt')
p1.write_text('I like playing guitar.')
content4 = p1.read_text()
print(content4)
```

```python
#python只能将字符串写入文本文件，如果要将数值数据存储到其中，必须用str()转化为字符串格式
#写入多行
content5 = "I love programing\n"
content5 += "I love playing guitar\n"
content5 += "I love dancing."
p1.write_text(content5)
content4 = p1.read_text()
print(content4)
```

```python
def show_content(file_name):
    path=Path(file_name)
    try:
        contents=path.read_text()
    except:
        print("the file dosen`t exist.")
    else:
        print(contents)
```

```python
file_name=['cat.txt','dog.txt']
for name in file_name:
    show_content(name)
```
