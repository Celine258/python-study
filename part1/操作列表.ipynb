```python
magicians=['Alice','David','Celine']
for magician in magicians:
    print(magician)
```

```python
#循环很重要
for magician in magicians:
    print(f"{magician},\tthat was a great trick!")
    print(f"I can`t wait to see you next trick,{magician}.\n")


print("Thank you everyone.That was a great magic show!")
```

### 创建数值列表

```python
for i in range(1,6):
    print(i)
```

```python
even_numbers=list(range(2,11,2))
print(even_numbers)
```

```python
squares=[]
for value in range(1,11,2):
    square=value**2
    squares.append(square)
print(squares)
```

```python
print(min(squares))
print(max(squares))
print(sum(squares))
```

```python
#列表推导式
squares_new=[value**2 for value in range(2,11,2)]
print(squares_new)
```
