```python
def make_alnum(singer,name):
    alnum = {'singer' : singer,'name' : name}
    return alnum
a = make_alnum('Mili','今天，明天')
print(a)
print(make_alnum('Stack','阳炎'))
```

```python
#传递任意数量的实参
def pizza_toppings(*toppings):#*的作用是让Python创建一个名为toppings的元组
    print(toppings)
pizza_toppings('1','2','3')
```

```python
#结合使用位置实参和任意数量的实参
def pizza(size,*toppings):
    print(size)
    for i in toppings:
        print(i)
pizza(1,'a','b','c')
```

```python
#**的作用就是创建一个字典，接受任意数量的关键词实参，也可以叫做键值对
def name(first,last,**info):
    print(first)
    print(last)
    for i,j in info.items():
        print(f"{i} : {j}")
name('Celine','Brando',location='princeton',field='phisics')
```

```python
def name(first,last,**info):
    info['first_name'] = first
    info['last_name'] = last
    return info
print(name('Celine','Brando',location='princeton',field='phisics'))
```
