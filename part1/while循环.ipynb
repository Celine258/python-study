```python
notice = "age < 3 free\n"
notice += "3 <= age <12 10$\n"
notice += "age >= 12 15$\n"
notice += "Plz enter your age.\n"
notice += "(enter 'quit' when you finish it.)\n"
age = ""
while age != 'quit':
    age = input(notice)
    age1 = int(age)
    if age1 < 3:
        print("free")
    elif age1 < 12:
        print("12$")
    else:
        print("15$")
```
