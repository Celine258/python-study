Python之禅

```python
#列表
fruit=['apple','banana','pear']
print(fruit)
print(fruit[0])
print(fruit[0].title())

massege=f"My favorite fruit is {fruit[0].title()}"
print(massege)
```

```python
#修改元素
fruit[0]='peach'
print(fruit[0].title())
```

```python
fruit.append('apple')
print(fruit)
```

```python
fruit.insert(1,'watermalon')
print(fruit)
```

```python
del fruit[0]
print(fruit)
```

```python
pop_fruit=fruit.pop()#无参数提取末尾元素
print(pop_fruit)
print(fruit)
```

```python
pop_fruit=fruit.pop(0)#有参数提取指定参数位置数
print(pop_fruit)
print(fruit)
```

```python
fruit=['apple','banana','pear','peach']
```

```python
fruit.remove('peach')
print(fruit)
```
