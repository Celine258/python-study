```python
class Dog:
    def __init__(self,name,age):#类中的函数称为方法，其中self是一个指向示例本身的引用，让实例能够访问类中的属性和方法
        self.name = name#self为前缀能让类中所有的方法调用
        self.age = age
    def sit(self):
        print(f"{self.name} is now sitting")
    def roll_over(self):
        print(f"{self.name} rolled over!")
```

```python
my_dog = Dog('Wille',6)
print(my_dog.name)
print(my_dog.age)
```

```python
my_dog.sit()
```

```python
my_dog.roll_over()
```

```python
class Restaurant:
    def __init__(self,restaurant_name,cusine_type):
        self.restaurant_name = restaurant_name
        self.cusine_type = cusine_type
    def describe_restaurant(self):
        print(f"Welcome to {self.restaurant_name}")
        print(f"We have {self.cusine_type} types food.")
    def open_restaurant(self):
        print("The restaurant is opening!!!")
```

```python
my_restaurant = Restaurant('King of King',18)
print(my_restaurant.restaurant_name)
print(my_restaurant.cusine_type)
my_restaurant.describe_restaurant()
my_restaurant.open_restaurant()
```

```python
#修改属性的值，直接修改，通过方法修改，通过法让值递增
```

```python
#继承，父类，子类
class New_restaurant(Restaurant):
    def __init__(self, restaurant_name, cusine_type,owner):
        super().__init__(restaurant_name, cusine_type)
        self.owner = owner#可以添加其他属性,也可以增加其他方法,同时也可以重写父类有的方法
    def open_restaurant(self):
        print("We close today.")
```

```python
new = New_restaurant('A',12,'Celine')
new.describe_restaurant()
new.open_restaurant()
```

```python
#继承示例
class IceCreamStand(Restaurant):
    def __init__(self, restaurant_name, cusine_type):
        super().__init__(restaurant_name, cusine_type)
        self.flavors = ['chocolate','strawberry','banana','mango']
    def show_flavors(self):
        print("We have these flavors:")
        for i in self.flavors:
            print(i)
```

```python
icecream = IceCreamStand('A',4)
icecream.show_flavors()
```

```python
class Die:
    def __init__(self,sides = 6):
        self.sides = sides
    def roll_die(self):
        for i in range(1,11):
            print(randint(1,self.sides))
```

```python
die = Die()
die.roll_die()
```
