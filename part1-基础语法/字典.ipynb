```python
favorite_language={
    'jen':['python','rust'],
    'sarah':'c',
    'edward':['rust','go'],
    'phil':['python','haskell']
}
for name,languaage in favorite_language.items():
    print(f"\n{name}`s favorite language is:")
    if len(languaage) == 1:
        print(languaage)
    else:
        for i in languaage:
            print(i)
```

```python
pets = {
    'ketyy':{
        'owner':'Alex',
        'age':2,
        'kind':'cat',
        'food':['fish','chiken'],
    },
    'Huang':{
        'owner':'Celine',
        'age':3,
        'kind':'dog',
        'food':'chiken',
    }
}
for name,info in pets.items():
    print(name)
    print(info['owner'])
    print(info['kind'])
    if len(info['food']) == 6:
        print(info['food'])
    else:
        for i in info['food']:
            print(i)
    print('\n')
print(len(pets['Huang']['food']))
```
