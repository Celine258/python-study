```python
#测试
class AnonymousSurvey:
    def __init__(self,question):
        self.question=question
        self.reponses=[]

    def show_quesion(self):
        print(self.question)

    def stored_response(self, new_response):
        self.reponses.append(new_response)

    def show_response(self):
        for response in self.reponses:
            print(response)

question = "Which fruit do you like?"
fruit_survey = AnonymousSurvey(question)
fruit_survey.show_quesion()
print("Enter 'q' to quit." )
while True:
    reponse = input('Fruit')
    if reponse == 'q':
        break
    fruit_survey.stored_response(reponse)

print("result")
fruit_survey.show_response()
#详情survey.py & test_survey.py
```
