```python
import json
with open("D:\Python学习\part2\地图\疫情.txt","r",encoding="utf-8") as f:
    data = f.read()
data_dict = json.loads(data)
province_data_dict = data_dict["areaTree"][0]["children"]
data_list = []
municipalities = {"北京", "上海", "天津", "重庆"}        # 直辖市 → XX市
autonomous_region = {"内蒙古", "西藏"} # 自治区 → XX自治区
hk_macao = {"香港", "澳门"}                              # 港澳，不加后缀
for province_data in province_data_dict:
    province_name = province_data["name"]
    province_comfirm = province_data["total"]["confirm"]
    if province_name in municipalities:
        province_name += "市"
    elif province_name in autonomous_region:
        province_name += "自治区"
    elif province_name in hk_macao:
        province_name += "特别行政区"
    elif province_name == "台湾":
        province_name += "省"
    elif province_name in {"广西"}:
        province_name += "壮族自治区"
    elif province_name in {"新疆"}:
        province_name += "维吾尔自治区"
    elif province_name in {"宁夏"}:
        province_name += "回族自治区"   
    else:
        # 其余普通省份
        province_name += "省"
    data_list.append((province_name,province_comfirm))
print(data_list)
```

```python
from pyecharts.charts import Map
from pyecharts import options as opts 
map = Map()
map.add("各省份新冠确诊人数",data_list,"china")
map.set_global_opts(
    title_opts=opts.TitleOpts(title="全国疫情地图"),
    visualmap_opts=opts.VisualMapOpts(
        is_show=True,
        is_piecewise=True,
        pieces=[
            {"min":1,"max":99,"label":"1-99人","color":"#CCFFFF"},
            {"min":100,"max":999,"label":"100-999人","color":"#FFFF99"},
            {"min":1000,"max":9999,"label":"1000-9999人","color":"#FF9966"},
            {"min":10000,"max":99999,"label":"10000-99999人","color":"#FF6666"},
            {"min":100000,"label":"100000+人","color":"#910000"},
        ]
    )
)
map.render("疫情地图.html")
```
