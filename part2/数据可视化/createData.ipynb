```python
import matplotlib.pyplot as plt
squares = [1,4,9,16,25]
input_values = [1,2,3,4,5]
fig, ax = plt.subplots()
ax.plot(input_values,squares,linewidth=3)

ax.set_title("Square Numbers",fontsize=24)
ax.set_xlabel("Numbers",fontsize=14)
ax.set_ylabel("Value of numbers",fontsize=14)
ax.tick_params(labelsize=14)

plt.show
```

```python
squares = [1,4,9,16,25]
input_values = [1,2,3,4,5]
fig, ax = plt.subplots()
ax.plot(input_values,squares,linewidth=3)

plt.style.use('seaborn-v0_8-white')
ax.set_title("Square Numbers",fontsize=24)
ax.set_xlabel("Numbers",fontsize=14)
ax.set_ylabel("Value of numbers",fontsize=14)
ax.tick_params(labelsize=14)

plt.show
```

```python
plt.style.use('seaborn-v0_8-ticks')
fig, ax = plt.subplots()
ax.scatter(2,4,s=200)
ax.set_title("Square Number",fontsize=24)
ax.set_xlabel("Numbers",fontsize=14)
ax.set_ylabel("Value of number",fontsize=14)
ax.tick_params(labelsize=14)

plt.show()
```

```python
x_values=[1,2,3,4,5]
y_values=[1,4,9,16,25]

fig, ax = plt.subplots()
ax.scatter(x_values,y_values,s=200)

ax.set_title("Square Number",fontsize=24)
ax.set_xlabel("Numbers" , fontsize=14)
ax.set_ylabel("Square of number",fontsize=14)
ax.tick_params(labelsize=14)

plt.show()
```

```python
x_values1=range(1,1001)
y_values1=[x**2 for x in x_values1]

fig, ax = plt.subplots()
ax.scatter(x_values1,y_values1,s=10,c=y_values1,cmap=plt.cm.Reds)
ax.set_title("Square Numbers",fontsize=24)
ax.set_xlabel("Numbers",fontsize=14)
ax.set_ylabel("Squares",fontsize=14)
ax.tick_params(labelsize=14)

ax.axis([0,1001,0,1_100_100])
plt.show()
```

```python
x_value2=range(0,5000)
y_value2=[x**3 for x in x_value2]

fig, ax =plt.subplots()
ax.set_title("The cube of numbers",fontsize=24)
ax.set_xlabel("Numbers",fontsize=14)
ax.set_ylabel("Cube",fontsize=14)
ax.tick_params(labelsize=14)
ax.scatter(x_value2,y_value2,c=y_value2,cmap=plt.cm.Reds,s=10)
plt.show()
```
