```python
import json
data = [{"name": "alice","age": 18},{"name":"celine","age":12},{"name":"dio","age":200}]
json_str = json.dumps(data)
print(type(json_str))
print(json_str)
```

```python
print(type(data))
```

```python
d = {"name":"周杰伦","addr":"台北"}
d_json = json.dumps(d,ensure_ascii=False)
print(type(d_json))
print(d_json)
```

```python
l = json.loads(json_str)
print(type(l))
print(l)
```
